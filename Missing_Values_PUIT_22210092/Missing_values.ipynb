```python
# STEP 1: Upload dataset
from google.colab import files
uploaded = files.upload()   # Select sensor_log.csv manually

# STEP 2: Load data
import pandas as pd
df = pd.read_csv("sensor_log.csv")  # filename only!
print("Original Data:")
display(df)

# STEP 3: Check missing values
print("\nMissing values in each column:")
print(df.isnull().sum())

# STEP 4: Create 4 different versions
# Forward Fill
df_forward_fill = df.fillna(method='ffill')

# Backward Fill
df_back_fill = df.fillna(method='bfill')

# Interpolation
df_interpolate = df.interpolate()

# Mean Fill
df_mean_fill = df.fillna(df.mean(numeric_only=True))


# STEP 5: Display results
print("\n--- Forward Fill ---")
display(df_forward_fill)

print("\n--- Backward Fill ---")
display(df_back_fill)

print("\n--- Interpolation ---")
display(df_interpolate)

print("\n--- Mean Fill ---")
display(df_mean_fill)

# STEP 6: Save EACH dataframe as CSV (optional but useful for CA)
df_forward_fill.to_csv("forward_fill.csv", index=False)
df_back_fill.to_csv("backward_fill.csv", index=False)
df_interpolate.to_csv("interpolated.csv", index=False)
df_mean_fill.to_csv("mean_fill.csv", index=False)

print("\nAll cleaned datasets saved as CSV files!")
```

Exercise 2

```python
# Copy of the original df
df_median = df.copy()

# Select only numeric columns
numeric_cols = df_median.select_dtypes(include="number").columns

# Median-based imputation
df_median[numeric_cols] = df_median[numeric_cols].fillna(
    df_median[numeric_cols].median()
)

# ---------- CREATE df_mean (this was missing) ----------
df_mean = df.copy()
df_mean[numeric_cols] = df_mean[numeric_cols].fillna(
    df_mean[numeric_cols].mean()
)
# -------------------------------------------------------

# Compare results
print("Mean-based imputation (df_mean):")
display(df_mean[numeric_cols].describe())

print("\nMedian-based imputation (df_median):")
display(df_median[numeric_cols].describe())
```

```python
import pandas as pd

# ✅ 1. Start from the original dataframe df
# If you already have df from sensor_log.csv, no need to reload it.
# df = pd.read_csv("sensor_log.csv")

# For this exercise, we'll treat the existing index as the time order
df_ts = df.copy().sort_index()

# ✅ 2. Create forward-fill and interpolation versions
df_ts_ffill = df_ts.ffill()          # Forward fill
df_ts_interp = df_ts.interpolate()   # Interpolation

# ✅ 3. Choose a small time range safely (e.g. first 10 rows or fewer)
end = min(10, len(df_ts))   # if you have fewer than 10 rows, it won't break

# Side-by-side comparison for temperature_c (you can also do for humidity, voltage)
comparison_temp = pd.DataFrame({
    "original_temp": df_ts["temperature_c"].iloc[0:end],
    "ffill_temp":    df_ts_ffill["temperature_c"].iloc[0:end],
    "interp_temp":   df_ts_interp["temperature_c"].iloc[0:end],
})

print("Temperature comparison (original vs forward fill vs interpolation):")
print(comparison_temp)

# ✅ Optional: also compare humidity and voltage the same way
comparison_hum = pd.DataFrame({
    "original_hum": df_ts["humidity_pct"].iloc[0:end],
    "ffill_hum":    df_ts_ffill["humidity_pct"].iloc[0:end],
    "interp_hum":   df_ts_interp["humidity_pct"].iloc[0:end],
})

print("\nHumidity comparison:")
print(comparison_hum)

comparison_volt = pd.DataFrame({
    "original_volt": df_ts["voltage_v"].iloc[0:end],
    "ffill_volt":    df_ts_ffill["voltage_v"].iloc[0:end],
    "interp_volt":   df_ts_interp["voltage_v"].iloc[0:end],
})

print("\nVoltage comparison:")
print(comparison_volt)
```
